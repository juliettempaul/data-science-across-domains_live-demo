```python
import tensorflow as tf
import numpy as np
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Dense, Flatten, Dropout
import PIL
import matplotlib.pyplot as plt
```

```python
import pathlib
path_to_train = "data-marginalia/page_images"
path_to_test = "data-marginalia/test_images"

train_early_print = pathlib.Path(path_to_train).with_suffix('')
test_early_print = pathlib.Path(path_to_test).with_suffix('')
```

```python
batch_size = 32
img_height = 300
img_width = 300
```

```python
train_early_print, validation_early_print = tf.keras.utils.image_dataset_from_directory(
  train_early_print,
  validation_split=0.2,
  subset="both",
  seed=123,
  image_size=(img_height, img_width),
  batch_size=batch_size)
```

```python
class_names = train_early_print.class_names
print(class_names)
```

```python
plt.figure(figsize=(10, 10))
for images, labels in train_early_print.take(1):
  for i in range(9):
    ax = plt.subplot(3, 3, i + 1)
    plt.imshow(images[i].numpy().astype("uint8"))
    plt.title(class_names[labels[i]])
    plt.axis("off")
```

```python
AUTOTUNE = tf.data.AUTOTUNE

train_early_print = train_early_print.cache().shuffle(1000).prefetch(buffer_size=AUTOTUNE)
validation_early_print = validation_early_print.cache().prefetch(buffer_size=AUTOTUNE)
```

```python
normalization_layer = tf.keras.layers.Rescaling(1./255)
```

```python
normalized_images = train_early_print.map(lambda x, y: (normalization_layer(x), y))
image_batch, labels_batch = next(iter(normalized_images))
first_image = image_batch[0]
print(np.min(first_image), np.max(first_image))
```

```python
layers = tf.keras.layers

data_augmentation = keras.Sequential(
  [
    layers.RandomFlip("horizontal",
                      input_shape=(img_height,
                                  img_width,
                                  3)),
    layers.RandomRotation(0.02),
  ]
)
```

```python
plt.figure(figsize=(10, 10))
for images, _ in train_early_print.take(1):
  for i in range(9):
    augmented_images = data_augmentation(images)
    ax = plt.subplot(3, 3, i + 1)
    plt.imshow(augmented_images[0].numpy().astype("uint8"))
    plt.axis("off")
```

```python
num_classes = len(class_names)

model = Sequential([
  data_augmentation,
  layers.Rescaling(1./255, input_shape=(img_height, img_width, 3)),
  layers.Conv2D(16, 3, padding='same', activation='relu'),
  layers.MaxPooling2D(),
  layers.Conv2D(32, 3, padding='same', activation='relu'),
  layers.MaxPooling2D(),
  layers.Conv2D(64, 3, padding='same', activation='relu'),
  layers.MaxPooling2D(),
  layers.Dropout(0.2),
  layers.Flatten(),
  layers.Dense(128, activation='relu'),
  layers.Dense(num_classes)
])
```

```python
model.compile(optimizer='adam',
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
              metrics=['accuracy'])
```

```python
epochs=5
history = model.fit(
  train_early_print,
  validation_data = validation_early_print,
  epochs=epochs
)
```

```python
acc = history.history['accuracy']
validation_acc = history.history['val_accuracy']

loss = history.history['loss']
validation_loss = history.history['val_loss']

epochs_range = range(epochs)

plt.figure(figsize=(8, 8))
plt.subplot(1, 2, 1)
plt.plot(epochs_range, acc, label='Training Accuracy')
plt.plot(epochs_range, validation_acc, label='Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend(loc='lower right')
plt.title('Training and Validation Accuracy')

plt.subplot(1, 2, 2)
plt.plot(epochs_range, loss, label='Training Loss')
plt.plot(epochs_range, validation_loss, label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend(loc='upper right')
plt.title('Training and Validation Loss')
plt.show()
```

## Visualizing what the model learned

**Grad-CAM**: which pixels the model actually weighted when making each decision, shown as a heatmap over the original image.

### Grad-CAM: where is the model looking?

Grad-CAM shows which pixels the model weighted most heavily to make its decision, as a heatmap laid over the original image. Broken into short steps below so each one can be run — and re-run — live in class. To try a different image, just change `example_path` in the second cell and re-run from there.

```python
# Find the model's last convolutional layer -- Grad-CAM reads what it detected
last_conv_layer_name = [layer.name for layer in model.layers if 'conv2d' in layer.name][-1]
print("Last convolutional layer:", last_conv_layer_name)
```

```python
# Pick one test image to explain -- change this path to try a different image
example_path = os.path.join(path_to_test1, os.listdir(path_to_test1)[0])

img = tf.keras.utils.load_img(example_path, target_size=(img_height, img_width))
img_array = tf.keras.utils.img_to_array(img)
img_array = tf.expand_dims(img_array, 0)  # add a batch dimension

plt.imshow(img_array[0].numpy().astype("uint8"))
plt.title("Original image")
plt.axis("off")
```

```python
# A version of the model that also reports the last conv layer's output.
# Rebuilt from a fresh Input so Keras traces every layer (including the
# nested data_augmentation block) -- same trained weights, just a graph
# Grad-CAM can read from.
inputs = keras.Input(shape=(img_height, img_width, 3))
x = inputs
for layer in model.layers:
    x = layer(x)
    if layer.name == last_conv_layer_name:
        conv_output = x

grad_model = tf.keras.models.Model(inputs, [conv_output, x])
```

```python
# Watch how much each part of that layer contributed to the predicted class
with tf.GradientTape() as tape:
    conv_output, predictions = grad_model(img_array)
    predicted_class = tf.argmax(predictions[0])
    class_score = predictions[:, predicted_class]

grads = tape.gradient(class_score, conv_output)
pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))

heatmap = conv_output[0] @ pooled_grads[..., tf.newaxis]
heatmap = tf.squeeze(heatmap)
heatmap = tf.maximum(heatmap, 0) / tf.math.reduce_max(heatmap)

plt.imshow(heatmap)
plt.title("Raw heatmap")
```

```python
# Overlay the heatmap on the original image
plt.imshow(img_array[0].numpy().astype("uint8"))
plt.imshow(heatmap, cmap="jet", alpha=0.4, extent=(0, img_width, img_height, 0))
plt.title(f"Grad-CAM -- predicted: {class_names[predicted_class]}")
plt.axis("off")
plt.show()
```
